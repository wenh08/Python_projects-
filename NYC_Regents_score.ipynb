```python
# Project question
#How did Geometry Regents exam scores differ between Staten Island and Brooklyn from 2015 to 2017?
# i choose Geometry because it is offered in year 2 and that's a better measure than freshmans
# now to subset a new dataframe so that i can focus on the columns that im looking at

regents_subset = df[['Year', 'School DBN', 'School Name','School Level', 'Regents Exam','Total Tested', 'Mean Score']]




```

```python
regents_subset.head()
```

```python


#Concatenate both DataFrames into one comparison dataset.
#Clean and investigate missing values, duplicates, and data types.
#Compare Geometry Regents performance by borough and year.
```

```python
#Find the NYC school directory to understand which DBNs belong to Staten Island and Brooklyn.
#Create a Staten Island DataFrame containing only those schools.


staten_island_subset = regents_subset[regents_subset['School DBN'].str[2]== "R"]

staten_island_subset
```

```python

#Create a Brooklyn DataFrame containing only Brooklyn schools.

brooklyn_subset = regents_subset[regents_subset['School DBN'].str[2] == 'K']
brooklyn_subset
```

```python
#Concatenate both DataFrames into one comparison dataset.
# We are using ignore_index = True because pandas keeps their original row indexes by default and here we get a fresh continuous index instead of repeating index naumbers from the original dataframe
# using pd.concat to combine and two dataframs
borough_comparison = pd.concat([staten_island_subset, brooklyn_subset], ignore_index= True)




borough_comparison
```

```python
borough_comparison['Borough'] = borough_comparison['School DBN'].str[2].map(
    {
    'R': 'Staten.Island',
    'K': 'Brooklyn',
    'M': 'Manhattan',
    'X': 'Bronx',
    'Q': 'Queens' }
)
```

```python
borough_comparison['Mean Score'].value_counts()
```

```python
borough_comparison.drop(borough_comparison[borough_comparison['Mean Score'] == 's'].index, axis=0, inplace= True)
```

```python
borough_comparison['Mean Score'].value_counts()
```

```python
borough_comparison['Mean Score'].dtype
# this column is still being consdiered as a object when we need it as numerical even after droping the row 's'
```

```python
borough_comparison['Mean Score'] = borough_comparison['Mean Score'].astype(float)

# Here we are changing the datatype as a float to make the column Mean score as numerical from object with .astype(float)
```

```python
borough_comparison['Mean Score'].dtype
# here we can see the data type has been changed to a float.
```

```python
borough_comparison.duplicated().sum()
# here we are checking for duplicated records because they can effect the mean
```

```python
borough_comparison[borough_comparison.duplicated(keep=False)].head(10)
#checking for duplicated records to make sure we dont remove legit exam records
```

```python
# Here i am cleaning the dataframe with .drop_duplicates(inplace=True) because i found that there is 1356 duplicated records

borough_comparison.drop_duplicates(inplace=True)



```

```python
#Here i am checking to see if any duplicates remain in my datafrane and the output states 0 duplicated within these column rows
borough_comparison.duplicated().sum()

```

```python
borough_comparison[borough_comparison['Borough'] == 'Brooklyn']
```

```python
# .isin() selects records belonging to either Geometry Exam
# groupby()['Year','Regents Exam'] puts these records by year and exam version
# .size() counts the records in each group
# Here i see that Common Core Geometry has the highest number of exams so i am selecting this exam type for my ananalysis


borough_comparison[borough_comparison['Regents Exam'].isin(['Geometry', 'Common Core Geometry'])].groupby(['Year', "Regents Exam"]).size()

```

```python
# Here i am calculating common core geometry exam scores, im grouping by year and borough, calculating the average of the mean scores for staten island and brooklyn only,
#  using unstack to put them into seprate columns
# we are using the mean score in order to calculae the average of the reported school exam averages, giving each record equal weight regardless of how many students took the exam
# this is because brooklyn schools has more student exam records than staten island schools

geometry_averages = borough_comparison[borough_comparison['Regents Exam'] == 'Common Core Geometry'].groupby(['Year', 'Borough'])['Mean Score'].mean().unstack()
geometry_averages

# Here we can see that the mean score within exams taken in brooklyn and staten island
# SI had a higher average score than brooklyn for students tested in Common Core Geometry for all of three years of 2015-1017





```

```python
# creating a bar plot
geometry_averages.plot(kind = 'bar',
color =['#34c0eb', '#ebd634']
                       )

plt.title('Common Core Geometry: Staten Island vs Brooklyn')
plt.xlabel('Year')
plt.ylabel('Average Mean Score')
plt.xticks(rotation = 0)
plt.legend(title = 'Borough')
plt.show()
```
