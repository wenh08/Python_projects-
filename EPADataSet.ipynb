```python
EPA_df = pd.read_csv("https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/mtcars.csv")
```

```python
EPA_df.head()
```

```python
EPA_df.describe()
```

```python
# Clean the dataset by naming the missing column name

EPA_df.rename(columns= {'Unnamed: 0': 'vehicle'}, inplace=True)
```

```python
# create a new dataframe that consists of a subset of the original
#columns (only mpg, hp, wt, and cyl)
# we need a list because we are selecting mulitiple columns

EDA_subset = EPA_df[['mpg', 'hp', 'wt', 'cyl']]

EDA_subset.head()
#checking to make sure the new dataframe exists with only these columns
```

```python
EDA_subset.rename(columns = {'cyl': 'cylinders',
                             'mpg': 'Miles per gallon', 'wt':'weight', 'hp': 'horse power'},
                   inplace = True)
```

```python
EDA_subset.head()
```

```python
#summary statistics for those columns (average mpg)

EDA_subset.describe()
```

```python
# summary statistics for those columns (average mpg)
print(EDA_subset['Miles per gallon'].mean())

```
